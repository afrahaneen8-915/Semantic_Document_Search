# Semantic Document Search using LangChain

## Domain: AI in Education

This project compares semantic search and keyword search across 20 educational documents using 15 queries. It uses LangChain, ChromaDB, and Sentence Transformers.

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

print("Libraries imported successfully!")
```

```python
results = pd.read_csv("../dataset/search_results.csv")
scores = pd.read_csv("../dataset/similarity_scores.csv")

print("Search Results:")
display(results.head())

print("Similarity Scores:")
display(scores.head())
```

```python
print("SEMANTIC DOCUMENT SEARCH - PROJECT SUMMARY")
print("------------------------------------------")

print("Total Documents:", 20)
print("Total Queries:", results["Query"].nunique())

print("Semantic Search Results:",
      len(results[results["Search_Type"] == "Semantic"]))

print("Keyword Search Results:",
      len(results[results["Search_Type"] == "Keyword"]))

print("Similarity Score Records:", len(scores))
```

```python
comparison = results.groupby("Search_Type").agg(
    Total_Results=("Document_ID", "count"),
    Unique_Documents=("Document_ID", "nunique")
)

display(comparison)
```

```python
semantic_results = results[results["Search_Type"] == "Semantic"]

display(
    semantic_results.groupby("Query").head(5)
)
```

```python
keyword_results = results[results["Search_Type"] == "Keyword"]

display(
    keyword_results.groupby("Query").head(5)
)
```

```python
print("SIMILARITY SCORE ANALYSIS")
print("-------------------------")

print("Average Similarity Score:",
      round(scores["Similarity_Score"].mean(), 4))

print("Highest Similarity Score:",
      round(scores["Similarity_Score"].max(), 4))

print("Lowest Similarity Score:",
      round(scores["Similarity_Score"].min(), 4))
```

## Conclusion

The Semantic Document Search project successfully processes 20 educational documents and 15 queries.

Semantic search uses sentence embeddings and cosine similarity to identify documents based on meaning and context. Keyword search identifies documents based on matching words.

The project compares both approaches and displays the top 5 results for each query. ChromaDB stores the document embeddings, while LangChain supports the document search process.

This project demonstrates how AI-powered semantic search can improve information retrieval in the education domain.
